```python
from tensorflow.keras.datasets import mnist
import pandas as pd
import matplotlib.pyplot as plt
```

```python
(X_train, y_train), (X_test, y_test) = mnist.load_data()
assert X_train.shape == (60000, 28, 28)
assert X_test.shape == (10000, 28, 28)
assert y_train.shape == (60000,)
assert y_test.shape == (10000,)
```

```python
print("x_train:", X_train.shape)
print("x_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
import numpy as np

#Um mehrere Layer aneinander zu reihen/stoppeln braucht es mit 'Sequential' einen "Container"
model = Sequential()

#Jetzt ist es möglich n viele Layer in den Container zu packen. Wir verwdenden bei diesem ausschließlich dense layer
model.add(Input(shape=(784,)))
model.add(Dense(64, activation="relu"))
model.add(Dense(10, activation="softmax"))

#Wenn die Schichten stehen kann man das Model übersetzen:
model.compile(
    optimizer="rmsprop",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```
