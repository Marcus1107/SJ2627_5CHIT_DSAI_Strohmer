```python
import pandas as pd

# Dataset laden
df = pd.read_csv('movie.csv')

# Dimensionalität ermitteln (Zeilen, Spalten)
print("Dimensionalität:", df.shape)
```

```python
# Erste Zeilen anzeigen 
df.head()
```

```python
# Spalte selektieren und Typ ausgeben
print(type(df['movie_title']))
```

```python
# Spalte mit Initialwert 0 hinzufügen
df['has_seen'] = 0
```

```python
# Summe der Likes bilden
like_cols = ['director_facebook_likes', 'actor_1_facebook_likes', 
             'actor_2_facebook_likes', 'actor_3_facebook_likes']

df['director_actor_facebook_likes'] = df[like_cols].sum(axis=1, skipna=False)



# Anzahl der Zeilen mit NaN in den relevanten Like-Spalten
nan_count = df[like_cols].isna().any(axis=1).sum()
print(f"Anzahl der Einträge mit NaN: {nan_count}")


# NaNs in den Like-Spalten durch 0 ersetzen und Summe neu berechnen
df[like_cols] = df[like_cols].fillna(0)
df['director_actor_facebook_likes'] = df[like_cols].sum(axis=1)
```

```python
# Statistische Übersicht (Median, Mittelwert, Quantile) anzeigen
df['budget'].describe()

# Alternativ: Perzentil bestimmen
percentile = (df['budget'] < 22000000).mean() * 100
print(f"22 Mio. liegt bei ca. {percentile:.1f}% aller Film-Budgets.")
```
