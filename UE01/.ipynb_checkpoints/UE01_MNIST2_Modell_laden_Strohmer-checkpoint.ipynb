```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.utils import to_categorical

MODEL_PATH = Path.cwd() / "mnist2_best_model.keras"
if not MODEL_PATH.exists():
    raise FileNotFoundError("Zuerst UE01_MNIST2_Strohmer.ipynb vollständig ausführen.")

print("TensorFlow-Version:", tf.__version__)
print("Modelldatei:", MODEL_PATH)
```

```python
loaded_model = tf.keras.models.load_model(MODEL_PATH)
loaded_model.summary()
```

## Geladenes Modell mit MNIST-Testdaten prüfen

```python
(_, _), (x_test_raw, y_test_raw) = tf.keras.datasets.mnist.load_data()
x_test = x_test_raw.reshape(-1, 784).astype("float32") / 255.0
y_test = to_categorical(y_test_raw, num_classes=10)

test_loss, test_accuracy = loaded_model.evaluate(x_test, y_test, verbose=0)
print(f"Test-Loss: {test_loss:.4f}")
print(f"Test-Accuracy: {test_accuracy:.4f} ({test_accuracy:.2%})")
```

```python
predictions = loaded_model.predict(x_test[:9], verbose=0)
predicted_digits = np.argmax(predictions, axis=1)

fig, axes = plt.subplots(3, 3, figsize=(8, 8))
for index, axis in enumerate(axes.flat):
    axis.imshow(x_test_raw[index], cmap="gray_r")
    correct = predicted_digits[index] == y_test_raw[index]
    color = "green" if correct else "red"
    axis.set_title(
        f"Vorhersage: {predicted_digits[index]} | Label: {y_test_raw[index]}",
        color=color,
    )
    axis.axis("off")

plt.suptitle("Vorhersagen des geladenen Modells")
plt.tight_layout()
plt.show()
```

Das Modell wurde erfolgreich aus der `.keras`-Datei geladen und kann ohne erneutes Training für Evaluation und Vorhersagen verwendet werden.
